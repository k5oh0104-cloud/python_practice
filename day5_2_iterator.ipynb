# 반복가능한 객체  Iterable

 1. 시퀀스 객체 ; 순서가 있고  인덱스로 접근이 가능한 객체
     list, tuple, str, range
 2. Mapping 객체  키(key) 값(value)으로 이루어짐
    dict(딕셔너리)

3. Itrator ; next로 하나씩 자료를 꺼내는 객체
   한번 소모하면 다시 처음으로 꺼낼수 없음

```python
str = "안녕하세요 만나서 반갑습니다"
for ch in str:
  print(ch, end=('  '))
```

```python
names = ['김범주', '임규현', '오경진', '김원준']


print(names)

for i in range(len(names)):
  print(names[i])

for data in names:
  print(data, end="  ")
```

```python
data = [10, 5, 3, 6, 20, 50, 15]
max = 0
min = 0
for i in range(len(data)):
  if data[i] > max:
    max = data[i]
min = data[0]

for i in range(len(data)):
  diff = 0
  diff = min - data[i]
  if diff > 0:
    min = data[i]
print(f'The maximum value of the list is: {max}, and the minimum value of the list is: {min}')
```

```python
"""
data=[1,4,2,6,50,40,32]

4의 배수의 자료만 누적해서 출력후 합산도 출력하시오

새로운 배열을 만들어서

new_data=[4,40,32]

합계 : 76
"""
total = 0
data = [1, 4, 2, 6, 50, 40, 32]
new_data = []
for i in data:
  if i%4 == 0:
    new_data.append(i)
    total += i
print(data)
print(f'4로 나눠지는 수들은 {new_data}입니다.')
print(f'총 합계는 {total}입니다.')
```

```python
data = [10, 12, 13, 41, 25, 46, 17]
#2의 배수만 추려오기
double_data = [ i for i in data if i%2 == 0 ]
print(double_data)
```

```python
numbers = [1, 2, 3, 4, 5]
result = [ i*5 if i%2 == 0 else i for i in numbers]
print(result)
```
